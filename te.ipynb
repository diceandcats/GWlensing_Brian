```python
import numpy as np
import matplotlib.pyplot as plt

mu0 = 4*np.pi*1e-7
a = 1.0      # solenoid radius
L = 5.0      # solenoid length
n = 100.0    # turns per unit length
I = 1.0      # current

Z = np.linspace(-1.5*L, 1.5*L, 2000)

B = 0.5*mu0*n*I*((Z+L/2)/np.sqrt(a**2+(Z+L/2)**2) - (Z-L/2)/np.sqrt(a**2+(Z-L/2)**2))

plt.plot(Z, B)
plt.axvline(+L/2, linestyle='--')
plt.axvline(-L/2, linestyle='--')
plt.xlabel("Z")
plt.ylabel("B(Z)")
# hide x and y ticks
plt.xticks([])
plt.yticks([])
plt.show()
```
